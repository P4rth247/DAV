```python
import numpy as np
import pandas as pd
```

```python
print("-----------NUMPY-------------")
marks = np.array([85,72,90,68,95,78])
print("Orginal Arrays:")
print(marks)
print("\n--------Numpy Indexing--------")
print("First mark:",marks[0])
print("Third mark:",marks[2])
print("Last element:",marks[-1])
```

```python
print("\n--------Numpy Slicing-----")
print("First three marks:",marks[0:3])
print("Marks from index 2 to 4:",marks[2:5])
print("Every second marks:",marks[::2])
print("Reverse:",marks[::-1])
      
```

```python
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])
C = A + B
print("Matrix A:")
print(A)
print("Matrix B:")
print(B)
print("Addition:")
print(C)
```

```python
a=np.array([25000,32000,28000,35000,40000,38000])
print("Sum:",np.sum(a))
print("Average:",np.mean(a))
print("Maximum:",np.max(a))
print("Minimum:",np.min(a))
print("Sales from third month:",a[2:])
print("Every second month sales:",a[0::2])
```

```python
data = {
    "Name": ["Rahul", "Priya", "Amit", "Sneha"],
    "Marks": [75, 90, 65, 95]
}
df = pd.DataFrame(data)
result = df[df["Marks"] > 80]
print(result)
```

```python
data = {
    "Name": ["John", "Alice", "Bob", "David"],
    "Marks": [85, 90, 75, 88]
}
df = pd.DataFrame(data)
average = df["Marks"].mean()
print("Average marks:", average)
```
